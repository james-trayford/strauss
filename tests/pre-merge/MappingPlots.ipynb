# Mapping plots and animation

How an `AudioFigure`'s data maps to sound, as a plot and as a video with the audio.

Uses made-up data and the synthesiser styles (`default`, `windy`), so nothing is downloaded. Saving the video needs `ffmpeg` installed.

```python
# on Colab, install the branch first
# %pip --quiet install git+https://github.com/james-trayford/strauss.git@plot-theme
```

```python
import numpy as np
import matplotlib.pyplot as plt
import IPython.display as ipd
from strauss.audio_figure import AudioFigure
```

A mock pulsating star: every 3.2 days it brightens and heats up quickly, then cools and fades.

```python
rng = np.random.default_rng(42)
days = np.linspace(0, 12, 600)
phase = (days / 3.2) % 1
brightness = 1 + 0.4*np.exp(-phase/0.15) + 0.02*rng.standard_normal(days.size)
temperature = 6000 + 1500*np.exp(-phase/0.25) + 40*rng.standard_normal(days.size)

fig, axes = plt.subplots(2, 1, sharex=True, figsize=(6, 4))
axes[0].plot(days, brightness); axes[0].set_ylabel('Brightness')
axes[1].plot(days, temperature); axes[1].set_ylabel('Temperature [K]')
axes[1].set_xlabel('Time [days]');
```

Brightness sets the pitch (`default` style, raised from its C2 to C3 with `notes`), and temperature sets the filter cutoff (`windy` style), so the star sounds brighter as it gets hotter.

```python
afig = AudioFigure()
afig.sonify(days, brightness, name='brightness', duration=12, notes=['C3'])
afig.sonify(days, temperature, style='windy', name='temperature', duration=12)
afig.notebook_display()
```

## Mapping plot

One column per sonification. Each panel shows the input data against the input time on the left and bottom axes, with the sound it maps to on the right and top axes.

```python
afig.plot_mapping(title='Pulsating star');
```

```python
afig.plot_mapping(title='Pulsating star', theme='light');
```

## Animation

A playhead runs over the plot in time with the audio, marking the value sounding in each panel, with the waveform below. The video is written with the figure's mixed audio.

```python
afig.animate_mapping(fname='pulsating_star.mp4')
ipd.Video('pulsating_star.mp4', embed=True, width=720)
```
