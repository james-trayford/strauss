# Mapping plots and animation

How an `AudioFigure`'s data maps to sound, as a plot and as a video with the audio.

Uses made-up data and the synthesiser styles (`default`, `windy`), so nothing is downloaded. Saving the video needs `ffmpeg` installed.

```python
# on Colab, install the branch under test (elsewhere, use your own install)
import sys
if 'google.colab' in sys.modules:
    %pip --quiet install git+https://github.com/james-trayford/strauss.git@note-aliases
```

```python
import numpy as np
import matplotlib.pyplot as plt
import IPython.display as ipd
from strauss.audio_figure import AudioFigure
```

A mock pulsating star: every 3.2 days it brightens and heats up quickly, then cools and fades.

```python
rng = np.random.default_rng(42)
days = np.linspace(0, 12, 600)
phase = (days / 3.2) % 1
brightness = 1 + 0.4*np.exp(-phase/0.15) + 0.02*rng.standard_normal(days.size)
temperature = 6000 + 1500*np.exp(-phase/0.25) + 40*rng.standard_normal(days.size)

fig, axes = plt.subplots(2, 1, sharex=True, figsize=(6, 4))
axes[0].plot(days, brightness); axes[0].set_ylabel('Brightness')
axes[1].plot(days, temperature); axes[1].set_ylabel('Temperature [K]')
axes[1].set_xlabel('Time [days]');
```

Brightness mapped to lowpass filter cutoff applied to white noise (the `'windy'` style)

```python
afig = AudioFigure()
afig.sonify(days, brightness, style='windy', name='brightness', duration=12)
afig.notebook_display()
```

## Mapping plot & Animation


### One Sonification

Each panel shows the input data against the input time on the left and bottom axes, with the sound it maps to on the right and top axes.

```python
afig.plot_mapping(title='Pulsating star');
```

```python
afig.plot_mapping(title='Pulsating star', theme='light');
```

When animated, a playhead moves across the plot in time with the audio, trackig the sonified parameters. The video is written with the figure's mixed audio.

```python
afig.animate_mapping(fname='pulsating_star_brightness.mp4')
ipd.Video('pulsating_star_brightness.mp4', embed=True, width=720)
```

### Multiple Sonifications

Let's make a two-sonification audio figure, now adding temperature mapped to pitch

```python
afig = AudioFigure()
afig.sonify(days, brightness, style='windy', name='brightness', duration=12)
afig.sonify(days, temperature, notes=['C3'], name='temperature', duration=12)
afig.notebook_display()
```

...and again animate. Sonifications are stacked as columns, with independently mapped parameters per sonification stacked as rows.

```python
afig.animate_mapping(fname='pulsating_star_full.mp4')
ipd.Video('pulsating_star_full.mp4', embed=True, width=720)
```
