# NaN handling examples

Examples of how NaN values are handled 

```python
import strauss
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 90, 400)
y =  np.sin(x/2)
z = 0.1*x**2 - x - 5 

idx = y.size//5
y[idx:idx+30] = np.nan

plt.plot(x,y)
plt.show()
```

```python
handle_type = 'silent' # this is the default, but can be changed to 'interpolate' 
```

### NaN-handled audio

Try **`default`** sonification of an Object - the region of `NaN`s are muted or interpolated 

```python
strauss.sonify(x, y, handle_nans=handle_type)
mapped = strauss.get_table()
fixed = strauss.get_fixed_table()
strauss.display()
print("\n\nMapped sources table:")
strauss.list_tables()
strauss.close()
```

Or for Events, simply skip `NaN`-carrying event sources, or interpolate based on surrounding values.

```python
strauss.sonify(x, -y, style='mallets', handle_nans=handle_type)
afig = strauss._get_current_figure()
strauss.display()
strauss.close()
```
